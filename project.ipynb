```python
from kaggle.api.kaggle_api_extended import KaggleApi

api = KaggleApi()
api.authenticate()

api.dataset_download_files(
    "najir0123/walmart-10k-sales-datasets",
    path="data",
    unzip=True
)

print("Downloaded")
```

```python
import pandas as pd
df = pd.read_csv("data/Walmart.csv",encoding_errors="ignore")
df.head()
#df.shape
```

```python
df.describe()
```

```python
df.info()
```

```python
# All duplicates 
#df.duplicated()
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
df.duplicated().sum()
```

```python
df.shape
```

```python
#missing or null values
df.isnull().sum()
```

```python
#dropping all rows with null or missing records
df.dropna(inplace=True)
df.isnull().sum()
```

```python
df.shape
```

```python
df.dtypes
```

```python
# Converting data type of unit_price to float
df['unit_price'] = df['unit_price'].str.replace('$', '').astype(float)
df.head(5)
```

```python
df.info()
```

```python
df.columns = df.columns.str.lower()
df.head(5)
```

```python
#creating a new column for total sales
df['total'] = df['unit_price'] * df['quantity']
df.head(5)
```

```python
df.to_csv("data/cleaned_walmart.csv", index=False)
```
